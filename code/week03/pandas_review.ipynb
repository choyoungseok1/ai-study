item_price 컬럼의 달러 표시 문자($)를 제거하고 float 타입으로 저장하여 new_price 컬럼에 저장하라


```python
df["new_price"] = df["item_price"].str.replace("$","").astype(float)

```

```python
df["choice_description"].isnull().sum()
```
