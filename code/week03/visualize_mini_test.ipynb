```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 깨짐 방지 (Windows)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 데이터 로드
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/pandas/main/chipo.csv')
df['new_price'] = df['item_price'].str.replace('$', '', regex=False).astype(float)


fig, axes = plt.subplots(2, 2, figsize=(15, 10))


```

```python
# 1) 인기 메뉴 top 10

top10 = df['item_name'].value_counts().head(10)
sns.barplot(x=top10.values, y=top10.index, ax=axes[0, 0])
axes[0, 0].set_title('인기 메뉴 Top 10')


```

```python
# 2) 가격 분포
sns.histplot(df['new_price'], ax=axes[0, 1])

```

```python
# 3) 메뉴별 평균 가격 top 10
avg_price = df.groupby('item_name')['new_price'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=avg_price.values, y=avg_price.index, ax=axes[1, 0])
```

```python
# 4) 수량 분포
sns.countplot(x='quantity', data=df, ax=axes[1, 1])
```

```python
# 2x2 subplot

plt.tight_layout()
plt.savefig('chipo_dashboard.png', dpi=100)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/pandas/main/chipo.csv')
df['new_price'] = df['item_price'].str.replace('$', '', regex=False).astype(float)

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

top10 = df['item_name'].value_counts().head(10)
sns.barplot(x=top10.values, y=top10.index, ax=axes[0, 0])
axes[0, 0].set_title('인기 메뉴 Top 10')

sns.histplot(df['new_price'], ax=axes[0, 1])
axes[0, 1].set_title('가격 분포')

avg_price = df.groupby('item_name')['new_price'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=avg_price.values, y=avg_price.index, ax=axes[1, 0])
axes[1, 0].set_title('메뉴별 평균 가격 Top 10')

sns.countplot(x='quantity', data=df, ax=axes[1, 1])
axes[1, 1].set_title('주문 수량 분포')

plt.tight_layout()
fig.savefig('chipo_dashboard.png', dpi=100, bbox_inches='tight')
plt.show()
```
