```python
from sklearn.datasets import load_diabetes
import torch
import numpy as np
data = (load_diabetes())
x = torch.FloatTensor(data.data)
y = torch.FloatTensor(data.target)
print(x.shape,y.shape)
```

```python
y=y.reshape(-1,1)
y.shape
```

```python
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(10, 1)
            
    def forward(self, x):
        return self.linear(x)
```

```python
model = MyModel()
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.001)
```

```python
x_mean = x.mean(dim=0)
x_std = x.std(dim=0)
x = (x - x_mean) / x_std
```

```python
for i in range(10000):
    optimizer.zero_grad()
    y_pred = model(x)
    loss = criterion(y_pred,y)
    loss.backward()
    optimizer.step()
    if i % 100 == 0:
        print(loss)
```

```python
from sklearn.datasets import load_breast_cancer

data = (load_breast_cancer())
x = torch.FloatTensor(data.data)
y = torch.FloatTensor(data.target)
y = y.reshape(-1,1)
print(x.shape,y.shape)

x_mean = x.mean(dim=0)
x_std = x.std(dim=0)
x = (x - x_mean) / x_std
```

```python
class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(30, 1)
            
    def forward(self, x):
        return torch.sigmoid(self.linear(x))
```

```python
model = MyModel()
criterion = nn.BCELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.001)
```

```python
for i in range(100000):
    optimizer.zero_grad()
    y_pred = model(x)
    loss = criterion(y_pred,y)
    loss.backward()
    optimizer.step()
    if i % 1000 == 0:
        print(loss)
```

```python
with torch.no_grad():
    pred = model(x)
    pred_label = (pred >= 0.5).float()
    acc = (pred_label == y).float().mean()
    print(f"Accuracy: {acc:.4f}")
```
