```python
import torch

x = torch.tensor(3.0, requires_grad=True)
y = x ** 2  # y = x²
y.backward()  # 미분 실행
print(x.grad)  # dy/dx = 2x = 6.0
```

```python
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)  # 입력 1개, 출력 1개
    
    def forward(self, x):
        return self.linear(x)
    
model = MyModel()
print(model)
print(list(model.parameters()))
```

```python
x_train = torch.tensor([[1.0],[2.0],[3.0],[4.0]])
y_train = torch.tensor([[3.0],[5.0],[7.0],[9.0]])  # y = 2x + 1

# 모델, 손실함수, 옵티마이저
model = MyModel()
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

```python
for i in range(1000):
    optimizer.zero_grad()
    y_pred = model(x_train)
    loss = criterion(y_pred,y_train)
    loss.backward()
    optimizer.step()
print(loss)
```

오늘 배운 거 정리:

Autograd: requires_grad=True → .backward() → .grad로 자동 미분
nn.Module: class로 모델 구조 정의 (레이어 + forward)
학습 루프 5줄: zero_grad → 예측 → loss → backward → step

이 5줄이 딥러닝 학습의 전부야. 앞으로 모델이 복잡해져도 이 구조는 똑같아.
