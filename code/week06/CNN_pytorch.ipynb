```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
```

```python
train_data = datasets.MNIST(root='./data', train=True, download=True, transform=transforms.ToTensor())
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_data = datasets.MNIST(root= "./data",train=False,download = True, transform=transforms.ToTensor())
test_loader = DataLoader(test_data,batch_size=64,shuffle= False)
```

```python
import torch.nn as nn

class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv2d1 = nn.Conv2d(1,32,3)
        self.conv2d2 = nn.Conv2d(32,64,3)
        self.Relu = nn.ReLU()
        self.maxpooling = nn.MaxPool2d(2)
        self.flatten = nn.Flatten()
        self.linear = nn.Linear(1600,10)
    def forward(self,x):
        x = self.conv2d1(x)
        x = self.Relu(x)
        x = self.maxpooling(x)
        x = self.conv2d2(x)
        x = self.Relu(x)
        x = self.maxpooling(x)
        x = self.flatten(x)
        return self.linear(x)
```

```python
model = CNN()
loss = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)
```

```python
for epoch in range(5):
    for images,labels in train_loader:
        predict = model(images)
        losscal = loss(predict,labels)
        optimizer.zero_grad()
        losscal.backward()
        optimizer.step()
        
```

```python
model.eval()
correct = 0
with torch.no_grad():
    for images, labels in test_loader:
        predict = model(images)
        predicted = predict.argmax(dim=1)
        correct += (predicted == labels).sum().item()
```

```python
print(f'정확도: {correct / len(test_data) * 100:.2f}%')
```

BatchNormalization + Drop Out 기법 추가

```python

import torch.nn as nn

class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv2d1 = nn.Conv2d(1,32,3)
        self.batchnorm1 = nn.BatchNorm2d(32)
        self.conv2d2 = nn.Conv2d(32,64,3)
        self.batchnorm2 = nn.BatchNorm2d(64)
        self.Relu = nn.ReLU()
        self.maxpooling = nn.MaxPool2d(2)
        self.flatten = nn.Flatten()
        self.dropout = nn.Dropout(0.5)
        self.linear = nn.Linear(1600,10)
    def forward(self,x):
        x = self.conv2d1(x)
        x = self.batchnorm1(x)
        x = self.Relu(x)
        x = self.maxpooling(x)
        x = self.conv2d2(x)
        x = self.batchnorm2(x)
        x = self.Relu(x)
        x = self.maxpooling(x)
        x = self.flatten(x)
        x = self.dropout(x)
        return self.linear(x)
```

```python
model = CNN()
loss = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)

for epoch in range(5):
    for images,labels in train_loader:
        predict = model(images)
        losscal = loss(predict,labels)
        optimizer.zero_grad()
        losscal.backward()
        optimizer.step()
        
```

```python
model.eval()
correct = 0
with torch.no_grad():
    for images, labels in test_loader:
        predict = model(images)
        predicted = predict.argmax(dim=1)
        correct += (predicted == labels).sum().item()
```

```python
print(f'정확도: {correct / len(test_data) * 100:.2f}%')
```
