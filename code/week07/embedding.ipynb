```python
import torch
import torch.nn as nn

torch.manual_seed(0)  # 재현용

vocab_size = 10   # 단어 10개라고 가정
embed_dim = 4     # 각 단어를 4차원 벡터로

embedding = nn.Embedding(vocab_size, embed_dim)

# 단어 인덱스 몇 개 넣어보기 (예: 1번, 5번, 3번 단어)
idx = torch.tensor([1, 5, 3])
vectors = embedding(idx)
print(vectors)
print(vectors.shape)   # 어떤 모양이 나올까? 직접 예측해보고 확인
```

```python
print(embedding.weight)            # 10 x 4 행렬 전체
print(embedding.weight[1])         # 1번 행
# 위에서 idx=1 넣었을 때 나온 vectors[0]이랑 같은지 비교
```

```python
optimizer = torch.optim.SGD(embedding.parameters(), lr=0.1)
target = torch.zeros(embed_dim)   # 1번 단어를 0벡터로 만들어보자

print("before:", embedding.weight[1])
for step in range(100):
    optimizer.zero_grad()
    vec = embedding(torch.tensor([1]))      # 1번 단어 벡터
    loss = ((vec - target) ** 2).mean()     # 목표와의 거리
    loss.backward()
    optimizer.step()
print("after:", embedding.weight[1])        # 0에 가까워졌나?
```
