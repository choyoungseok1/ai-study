## 개념 요약
- 원-핫: 차원 큼 + 단어 간 관계 못 담음(서로 직교).
- 임베딩: 저차원 dense 벡터, 의미 비슷하면 가까움. (king-man+woman≈queen)
- 분포 가설: 비슷한 문맥에 나오는 단어는 비슷한 의미 → CBOW/Skip-gram으로 학습.
- **`nn.Embedding` = 학습되는 행렬에서 인덱스로 행을 꺼내는 lookup table.**

```python
import torch
import torch.nn as nn

vocab_size, dim = 10, 4
emb = nn.Embedding(vocab_size, dim)

idx = torch.tensor([1, 5, 5, 2])
print(emb(idx))           # 인덱스 → 행 벡터 (룩업)
print(emb.weight.shape)   # (10, 4) = 학습되는 임베딩 행렬 전체

# 같은 인덱스는 같은 벡터:
print(torch.equal(emb(torch.tensor([5])), emb(torch.tensor([5]))))  # True
```
